```python
# assign the activity to the compounds
import numpy as np
df_bioactivity_status = final_unified_df.copy()

le_relations = ['=', '<=', '<']
ge_relations = ['=', '>=', '>']

unit = df_bioactivity_status['compound_standard_unit'].astype(str).str.strip()
rel  = df_bioactivity_status['compound_standard_relation'].astype(str).str.strip()
val  = df_bioactivity_status['compound_standard_value']

active_uM = (
    (unit.str.lower() == 'um') &
    (val < 10))

active_percent = (
    (unit == '%') &
    (rel.isin(ge_relations)) &
    (val >= 50))

df_bioactivity_status['activity'] = np.where(
    active_uM | active_percent,
    'active',
    'inactive')

df_bioactivity_status
```

```python
# m# keeping target names and compound smiles and their activity
all_df = df_bioactivity_status[['target_chembl_id',"compound_chembl_id", 'nonisomeric_smiles',"activity", "target_name", "target_type", "target_organism"]]
all_df
```

```python
# Apply separate filtering: min 5 active per target, min 5 inactive per target
active_per_target = all_df[all_df['activity'] == 'active'].groupby('target_chembl_id')['nonisomeric_smiles'].nunique()
inactive_per_target = all_df[all_df['activity'] == 'inactive'].groupby('target_chembl_id')['nonisomeric_smiles'].nunique()

valid_active_targets = set(active_per_target[active_per_target >= 5].index)
valid_inactive_targets = set(inactive_per_target[inactive_per_target >= 5].index)

# Keep only rows where the target passes the filter for its activity type
filtered_df = all_df[
    ((all_df['activity'] == 'active') & (all_df['target_chembl_id'].isin(valid_active_targets))) |
    ((all_df['activity'] == 'inactive') & (all_df['target_chembl_id'].isin(valid_inactive_targets)))
]

print(f"Before filtering: {len(all_df):,} rows")
print(f"After filtering:  {len(filtered_df):,} rows")
print(f"\nActive targets:   {filtered_df[filtered_df['activity'] == 'active']['target_chembl_id'].nunique():,}")
print(f"Inactive targets: {filtered_df[filtered_df['activity'] == 'inactive']['target_chembl_id'].nunique():,}")
print(f"Total targets:    {filtered_df['target_chembl_id'].nunique():,}")
print(f"Total compounds:  {filtered_df['nonisomeric_smiles'].nunique():,}")

filtered_df
```

```python
# Load the protein class lookup
protein_class_df = pd.read_csv('protein_class_chembl_35.csv')

# Sanity check: any duplicate target IDs in the lookup?
dup_count = protein_class_df['target_chembl_id'].duplicated().sum()
if dup_count > 0:
    print(f"WARNING: {dup_count} duplicate target_chembl_id rows in protein_class_chembl_35.csv. Keeping first.")
    protein_class_df = protein_class_df.drop_duplicates(subset='target_chembl_id', keep='first')

print(f"Protein class lookup: {len(protein_class_df):,} unique targets")

# Merge into filtered_df
filtered_df = filtered_df.merge(
    protein_class_df[['target_chembl_id', 'protein_class_L1']],
    on='target_chembl_id',
    how='left')

# Check coverage
missing = filtered_df['protein_class_L1'].isna().sum()
total = len(filtered_df)
unique_missing_targets = filtered_df.loc[filtered_df['protein_class_L1'].isna(), 'target_chembl_id'].nunique()

print(f"Rows with no protein class: {missing:,} / {total:,} ({100*missing/total:.2f}%)")
print(f"Unique targets with no protein class: {unique_missing_targets:,}")

# Fill missing protein classes with 'unclassified'
filtered_df['protein_class_L1'] = filtered_df['protein_class_L1'].fillna('unclassified')

filtered_df.head()
```

```python
# Build TARGETCLASSIFICATION file: CHEMBL_ID, CLASS, ORGANISM, TYPE
target_classification = (
    filtered_df[['target_chembl_id', 'protein_class_L1', 'target_organism', 'target_type']]
    .drop_duplicates(subset='target_chembl_id')
    .rename(columns={
        'target_chembl_id': 'CHEMBL_ID',
        'protein_class_L1': 'CLASS',
        'target_organism': 'ORGANISM',
        'target_type': 'TYPE'
    })
    .sort_values('CHEMBL_ID')
    .reset_index(drop=True))

# Sanity check: each CHEMBL_ID should map to one consistent (class, organism, type)
for col_orig, col_new in [('protein_class_L1', 'CLASS'),
                          ('target_organism', 'ORGANISM'),
                          ('target_type', 'TYPE')]:
    inconsistent = filtered_df.groupby('target_chembl_id')[col_orig].nunique()
    n_bad = (inconsistent > 1).sum()
    if n_bad > 0:
        print(f"WARNING: {n_bad} target IDs have multiple values for {col_new}. Keeping first.")

target_classification.to_csv('PPB4_TARGETCLASSIFICATION.txt', sep='\t', index=False)
print(f"Wrote {len(target_classification):,} targets to PPB4_TARGETCLASSIFICATION.txt")
print(target_classification.head())
```

```python
import os
import numpy as np
import pandas as pd
from rdkit import Chem
from rdkit.Chem import rdMolDescriptors
from rdkit.Chem.MolStandardize import charge
from tensorflow.keras.models import load_model

# ----------------------------------------------------------------------------
# Paths
# ----------------------------------------------------------------------------
MODELS_DIR = "models"
FP_BITS = 4096

# ----------------------------------------------------------------------------
# Load AtomPair active model + labels + metadata
# ----------------------------------------------------------------------------
model = load_model(
    os.path.join(MODELS_DIR, "ppb4_atompair_active_full_model.h5"),
    compile=False,
)
print(f"AtomPair active model output dim: {model.output_shape[-1]}")

with open(os.path.join(MODELS_DIR, "PPB4_ACTIVE_DNNTARLABELS.txt")) as f:
    active_labels = [line.strip() for line in f.readlines()[1:] if line.strip()]
print(f"Loaded {len(active_labels):,} active target labels")

assert model.output_shape[-1] == len(active_labels), \
    "AtomPair model output dim doesn't match active label count"

target_details = pd.read_csv(os.path.join(MODELS_DIR, "PPB4_TARGETSDETAILS.txt"), sep="\t")
target_class   = pd.read_csv(os.path.join(MODELS_DIR, "PPB4_TARGETCLASSIFICATION.txt"), sep="\t")
id_to_name     = target_details.set_index("CHEMBL_ID")["PREF_NAME"].to_dict()
id_to_class    = target_class.set_index("CHEMBL_ID")["CLASS"].to_dict()
id_to_organism = target_class.set_index("CHEMBL_ID")["ORGANISM"].to_dict()
id_to_type     = target_class.set_index("CHEMBL_ID")["TYPE"].to_dict()

# ----------------------------------------------------------------------------
# SMILES preprocessing
# ----------------------------------------------------------------------------
_uncharger = charge.Uncharger()

def preprocess_smiles(smi):
    mol = Chem.MolFromSmiles(smi)
    if mol is None:
        return None
    frags = Chem.GetMolFrags(mol, asMols=True, sanitizeFrags=False)
    largest = max(frags, key=lambda m: m.GetNumAtoms())
    cleaned = Chem.MolToSmiles(largest, isomericSmiles=False)
    mol = Chem.MolFromSmiles(cleaned)
    if mol is None:
        return None
    mol = _uncharger.uncharge(mol)
    return Chem.MolToSmiles(mol, isomericSmiles=False)

# ----------------------------------------------------------------------------
# AtomPair fingerprint — matches training script exactly
# ----------------------------------------------------------------------------
def calc_atompair(smi):
    mol = Chem.MolFromSmiles(smi)
    if mol is None:
        return None
    fp = rdMolDescriptors.GetHashedAtomPairFingerprintAsBitVect(mol, nBits=4096)
    return np.array(fp, dtype=np.int8)

# ----------------------------------------------------------------------------
# Run prediction
# ----------------------------------------------------------------------------
query_smi = "c3ccc(CN2CCCC[C@H]1NCC[C@H]12)cc3"
print(f"\nQuery SMILES (raw):       {query_smi}")

cleaned = preprocess_smiles(query_smi)
print(f"Query SMILES (cleaned):   {cleaned}")

fp = calc_atompair(cleaned)
X = fp.reshape(1, -1).astype(np.float32)
print(f"Fingerprint shape:        {X.shape}, sum={int(fp.sum())} bits set")

preds = model.predict(X, verbose=0)[0]
print(f"Predictions shape:        {preds.shape}")
print(f"Prediction range:         min={preds.min():.4f}, max={preds.max():.4f}")

# Top 10
top_idx = preds.argsort()[-10:][::-1]
rows = []
for rank, idx in enumerate(top_idx, start=1):
    tid = active_labels[idx]
    rows.append({
        "rank":        rank,
        "target_id":   tid,
        "target_name": id_to_name.get(tid, "Unknown"),
        "P(active)":   round(float(preds[idx]), 4),
        "class":       id_to_class.get(tid, "Unknown"),
        "type":        id_to_type.get(tid, "Unknown"),
        "organism":    id_to_organism.get(tid, "Unknown"),
    })

top10 = pd.DataFrame(rows)
print("\nTop 10 AtomPair active predictions:")
print(top10.to_string(index=False))
```

```python
import pandas as pd
from rdkit import Chem
from tqdm import tqdm

# ----------------------------------------------------------------------------
# Inputs — edit CSV_PATH if your file is elsewhere
# ----------------------------------------------------------------------------
CSV_PATH = "chebml_35_full_data.csv"
QUERY    = "c3ccc(CN2CCCC[C@H]1NCC[C@H]12)cc3"

# ----------------------------------------------------------------------------
# Canonicalize query
# ----------------------------------------------------------------------------
query_mol = Chem.MolFromSmiles(QUERY)
query_canonical_iso    = Chem.MolToSmiles(query_mol, isomericSmiles=True)
query_canonical_noniso = Chem.MolToSmiles(query_mol, isomericSmiles=False)
print(f"Query (raw):       {QUERY}")
print(f"Query (canonical): {query_canonical_iso}")
print(f"Query (no stereo): {query_canonical_noniso}")

# ----------------------------------------------------------------------------
# Detect SMILES column
# ----------------------------------------------------------------------------
peek = pd.read_csv(CSV_PATH, nrows=5)
print(f"\nColumns: {peek.columns.tolist()}")

smiles_candidates = [c for c in peek.columns if "smiles" in c.lower()]
if not smiles_candidates:
    raise SystemExit(f"No SMILES column found. Columns: {peek.columns.tolist()}")
SMILES_COL = smiles_candidates[0]
print(f"Using SMILES column: {SMILES_COL}\n")

# ----------------------------------------------------------------------------
# Stream and search
# ----------------------------------------------------------------------------
CHUNK = 200_000
matches = []
total_rows = 0

for chunk in tqdm(pd.read_csv(CSV_PATH, chunksize=CHUNK), desc="Scanning"):
    total_rows += len(chunk)

    # Fast: literal string match against any of the three forms
    smi_strs = chunk[SMILES_COL].astype(str)
    hit_mask = smi_strs.eq(QUERY) | smi_strs.eq(query_canonical_iso) | smi_strs.eq(query_canonical_noniso)
    if hit_mask.any():
        matches.append(chunk[hit_mask])
        continue

    # Slower fallback: canonicalize every SMILES in the chunk and compare
    canon = []
    for s in smi_strs:
        m = Chem.MolFromSmiles(s)
        canon.append(Chem.MolToSmiles(m, isomericSmiles=False) if m else None)
    canon = pd.Series(canon, index=chunk.index)
    canon_mask = canon == query_canonical_noniso
    if canon_mask.any():
        matches.append(chunk[canon_mask])

print(f"\nScanned {total_rows:,} rows.")

if not matches:
    print("\n  ❌  Compound NOT found in chembl_35_full_data.csv")
else:
    hits = pd.concat(matches, ignore_index=True)
    print(f"\n  ✅  Found {len(hits)} matching row(s):\n")
    print(hits.to_string(index=False))
```
